```python
X = np.array([[1, 82, 26],
             [1, 45, 49],
             [1, 71, 76],
             [1, 22, 37],
             [1, 29, 40],
             [1, 9, 0],
             [1, 12, 2],
             [1, 18, 10],
             [1, 24, 92]])

y = np.array([71, 54, 43, 45, 21, 11, 30, 45, 10]).T

B = np.linalg.inv(X.T @ X) @ X.T @ y
B
```

```python
y_hat = X @ B
y_hat
```

```python
resid = y - y_hat
resid
```

```python
RSS = np.sum(resid**2) / (len(y) - X.shape[1])
RSS
```

```python
var_covar = RSS * np.linalg.inv(X.T @ X)
var_covar
```

```python
sd_b2 = np.sqrt(var_covar[2][2])
sd_b2
```
