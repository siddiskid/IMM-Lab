```python
import mne
import os
import os.path as op
import matplotlib as plt
import nibabel

# define subjects_dir (where freesurfer subjects folder is) and subject number
s = '002'
sub = f'vml_meg_{s}'
subj = f'vml_mri_{s}' 
meg_path =  '/users/mgarson.stu/.anaconda/VML/MEG'
subjects_dir =  '/users/mgarson.stu/.anaconda/VML/MRI'
```

```python
mne.viz.set_3d_backend('pyvistaqt')
mne.gui.coregistration(subjects_dir=subjects_dir, subject=f'vml_mri_{s}')
```

```python
# make boundary element model (BEM) model & solution 
model = mne.make_bem_model(subject=subj, ico=5, conductivity=(0.3,),
                           subjects_dir=subjects_dir)
bem_sol = mne.make_bem_solution(model)

mne.write_bem_surfaces(op.join(subjects_dir, subj, 'bem', f'{s}-bem-model.fif'), model, overwrite=True)

# save bem solution file
mne.write_bem_solution(op.join(subjects_dir, subj, 'bem', 
                               f'{s}-bem-sol.fif'), bem_sol, overwrite=True)
```

```python
# plot bem
mne.viz.plot_bem(subject=subj, subjects_dir=subjects_dir)
```
