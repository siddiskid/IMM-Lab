```python
i = 0 
for s in adult_subs:
    sub = f'vml_meg_{s}'
    subj = f'vml_mri_{s}'
    meg_data = op.join(meg_path, sub, 'vml_meg_%s_tsss_ica_rot_ave.fif' %s)
    meg_data2 = op.join(meg_path, sub, 'vml_meg_%s_tsss_ica_run_ave.fif' %s)

    evoked = mne.read_evokeds(meg_data, proj=True, kind='average', verbose=False) #rot 
    evoked2 = mne.read_evokeds(meg_data2, proj=True, kind='average', verbose=False) #run

    bem_dir = op.join(mri_path, subj, 'bem')
    fname_aseg = op.join(mri_path, subj, 'mri', 'aseg.mgz')
    fname_model = op.join(bem_dir, f"{s}-bem-model.fif")
    fname_bem = op.join(bem_dir, f"{s}-bem-sol.fif")
    fname_trans = op.join(meg_path, sub, f'{sub}-trans.fif')
    fname_fwd = op.join(meg_path, sub, f'{s}-cerebellar-fw-solution-run.fif')
    erm_noise_cov = mne.read_cov(op.join(meg_path, f'vml_meg_{s}', f'{s}_noise-cov.fif'))
    rot_epochs = mne.read_epochs(op.join('VML', 'MEG', sub, f'{sub}_rot_epochs-epo.fif'), verbose=False)
    run_epochs = mne.read_epochs(op.join('VML', 'MEG', sub, f'{sub}_run_epochs-epo.fif'), verbose=False)

    if rot_epochs.baseline == run_epochs.baseline and evoked[0].baseline == evoked2[0].baseline: 
        print (f'Baselines are all the same:{rot_epochs.baseline}')
    else:
        print('Baselines are Different')
        
    el_epos = run_epochs['run2']
    ll_epos = run_epochs['run3']
    fwd = mne.read_forward_solution(fname_fwd, verbose=False)
    src = fwd['src']
    run_filters = mne.beamformer.read_beamformer(op.join(meg_path, sub, f'{s}-run_filters-lcmv.h5'))

    #Early_learning 
    el_stcs = []
    for i in range(len(el_epos)):
        stc = mne.beamformer.apply_lcmv_epochs(el_epos[i], run_filters, return_generator=True, verbose=False)
        el_stcs.append(list(stc))

    #Late Learning
    ll_stcs = []
    for i in range(len(ll_epos)):
        stc1 = mne.beamformer.apply_lcmv_epochs(ll_epos[i], run_filters, return_generator=True, verbose=False)
        ll_stcs.append(list(stc1))

    first_10_el_stcs = el_stcs[:10]
    last_10_ll_stcs = ll_stcs[19:29]

    vol_labels = mne.get_volume_labels_from_src(src, subject=subj, subjects_dir=mri_path)
    surf_labels = mne.read_labels_from_annot(subject=subj, parc='aparc', subjects_dir=mri_path)
    surf_labels = [label for label in surf_labels if label.name == 'postcentral-lh']
    times = el_stcs[0][0].times

    el_rh_tcs = []
    el_lh_tcs = []
    ll_rh_tcs = []
    ll_lh_tcs = []

    motor_el_lh_tcs = []
    motor_ll_lh_tcs = [] 


    for i in range(len(el_stcs)):
        #Extract the timecourses from early/late learning
        el_tcs = mne.extract_label_time_course(el_stcs[i], src=src, labels=vol_labels, mode='max', verbose=False)[:4]

        el_cerebellum_rh = np.mean((el_tcs[0][0], el_tcs[0][1]), axis=0)
        el_cerebellum_lh = np.mean((el_tcs[0][2], el_tcs[0][3]), axis=0)

        el_rh_tcs.append(el_cerebellum_rh)
        el_lh_tcs.append(el_cerebellum_lh)

        motor_el_tc = mne.extract_label_time_course(el_stcs[i], labels=surf_labels, src=src, mode='max')[0]
        motor_el_lh_tcs.append(motor_el_tc)

    for i in range(len(ll_stcs)):

        #Extract the timecourses from early/late learning
        ll_tcs = mne.extract_label_time_course(ll_stcs[i], src=src, labels=vol_labels, mode='max', verbose=False)[:4]

        #Average the Lh/Rh
        ll_cerebellum_rh = np.mean((ll_tcs[0][0], ll_tcs[0][1]), axis=0)
        ll_cerebellum_lh = np.mean((ll_tcs[0][2], ll_tcs[0][3]), axis=0)

        #Keep Track of the timecourses from each epoch 
        ll_rh_tcs.append(ll_cerebellum_rh)
        ll_lh_tcs.append(ll_cerebellum_lh)

        motor_ll_tc = mne.extract_label_time_course(ll_stcs[i], labels=surf_labels, src=src, mode='max')[0]
        motor_ll_lh_tcs.append(motor_ll_tc)

    #Shape of data is now (num_epochs x timepoints) (29 x 1701)
    el_rh_avg = np.mean(el_rh_tcs, axis=0)
    el_lh_avg = np.mean(el_lh_tcs, axis=0)
    ll_rh_avg = np.mean(ll_rh_tcs, axis=0)
    ll_lh_avg = np.mean(ll_lh_tcs, axis=0)

    #Motor Regions
    motor_el_lh_avg = np.mean(motor_el_lh_tcs, axis=(0,1))
    motor_ll_lh_avg = np.mean(motor_ll_lh_tcs, axis=(0,1))
    motor_el_lh_baseline = np.mean(motor_el_lh_avg[0:200])
    motor_ll_lh_baseline = np.mean(motor_ll_lh_avg[0:200])

    #First 10 el/ last 10 LL
    first_10_el_rh = np.mean(el_rh_tcs[:10], axis=0)
    first_10_el_lh = np.mean(el_lh_tcs[:10], axis=0)
    last_10_ll_rh = np.mean(ll_rh_tcs[19:29], axis=0)
    last_10_ll_lh = np.mean(ll_lh_tcs[19:29], axis=0)

    #Normalize WRT the baseline
    el_rh_baseline = np.mean(el_rh_avg[0:200])
    el_lh_baseline = np.mean(el_lh_avg[0:200])
    ll_rh_baseline = np.mean(ll_rh_avg[0:200])
    ll_lh_baseline = np.mean(ll_lh_avg[0:200])
    b1 = np.mean(first_10_el_rh[0:200])
    b2 = np.mean(first_10_el_lh[0:200])
    b3 = np.mean(last_10_ll_rh[0:200])
    b4 = np.mean(last_10_ll_lh[0:200])

    for i in range(len(el_rh_avg)):
        el_rh_avg[i] = el_rh_avg[i] - el_rh_baseline
        el_lh_avg[i] = el_lh_avg[i] - el_lh_baseline
        ll_rh_avg[i] = ll_rh_avg[i] - ll_rh_baseline
        ll_lh_avg[i] = ll_lh_avg[i] - ll_lh_baseline
        first_10_el_rh[i] = first_10_el_rh[i] - b1
        first_10_el_lh[i] = first_10_el_lh[i] - b2 
        last_10_ll_rh[i] = last_10_ll_rh[i] - b3
        last_10_ll_lh[i] = last_10_ll_lh[i] - b4

    for i in range(len(motor_el_lh_avg)):
        motor_el_lh_avg[i] = motor_el_lh_avg[i] - motor_el_lh_baseline
        motor_ll_lh_avg[i] = motor_ll_lh_avg[i] - motor_ll_lh_baseline

    sigma = 10  # Adjust sigma for more or less smoothing

    #Smooth the TC before finding the peaks: 
    el_rh_avg_smooth = gaussian_filter1d(el_rh_avg, sigma=sigma)
    el_lh_avg_smooth = gaussian_filter1d(el_lh_avg, sigma=sigma)
    ll_rh_avg_smooth = gaussian_filter1d(ll_rh_avg, sigma=sigma)
    ll_lh_avg_smooth = gaussian_filter1d(ll_lh_avg, sigma=sigma)
    first_10_el_rh_smooth = gaussian_filter1d(first_10_el_lh, sigma=sigma)
    first_10_el_lh_smooth = gaussian_filter1d(first_10_el_rh, sigma=sigma)
    last_10_ll_rh_smooth = gaussian_filter1d(last_10_ll_rh, sigma=sigma)
    last_10_ll_lh_smooth = gaussian_filter1d(last_10_ll_lh, sigma=sigma)

    motor_el_lh_avg_smooth = gaussian_filter1d(motor_el_lh_avg, sigma=sigma)
    motor_ll_lh_avg_smooth = gaussian_filter1d(motor_ll_lh_avg, sigma=sigma)

    # Compute dynamic thresholds and find peaks
    el_rh_thresh = (max(el_rh_avg_smooth[600:1000]) - min(el_rh_avg_smooth[600:1000])) / 2
    el_rh_peaks, _ = mne.preprocessing.peak_finder(el_rh_avg_smooth[600:1000], thresh=el_rh_thresh)
    el_rh_peaks = el_rh_peaks + 600

    el_lh_thresh = (max(el_lh_avg_smooth[600:1000]) - min(el_lh_avg_smooth[600:1000])) / 2
    el_lh_peaks, _ = mne.preprocessing.peak_finder(el_lh_avg_smooth[600:1000], thresh=el_lh_thresh)
    el_lh_peaks = el_lh_peaks + 600

    ll_rh_thresh = (max(ll_rh_avg_smooth[600:1000]) - min(ll_rh_avg_smooth[600:1000])) / 2
    ll_rh_peaks, _ = mne.preprocessing.peak_finder(ll_rh_avg_smooth[600:1000], thresh=ll_rh_thresh)
    ll_rh_peaks = ll_rh_peaks + 600

    ll_lh_thresh = (max(ll_lh_avg_smooth[600:1000]) - min(ll_lh_avg_smooth[600:1000])) / 2
    ll_lh_peaks, _ = mne.preprocessing.peak_finder(ll_lh_avg_smooth[600:1000], thresh=ll_lh_thresh)
    ll_lh_peaks = ll_lh_peaks + 600

    first_10_el_rh_thresh = (max(first_10_el_rh_smooth[600:1000]) - min(first_10_el_rh_smooth[600:1000])) / 2
    first_10_el_rh_peaks, _ = mne.preprocessing.peak_finder(first_10_el_rh_smooth[600:1000], thresh=first_10_el_rh_thresh)
    first_10_el_rh_peaks = first_10_el_rh_peaks + 600

    first_10_el_lh_thresh = (max(first_10_el_lh_smooth[600:1000]) - min(first_10_el_lh_smooth[600:1000])) / 2
    first_10_el_lh_peaks, _ = mne.preprocessing.peak_finder(first_10_el_lh_smooth[600:1000], thresh=first_10_el_lh_thresh)
    first_10_el_lh_peaks = first_10_el_lh_peaks + 600

    last_10_ll_rh_thresh = (max(last_10_ll_rh_smooth[600:1000]) - min(last_10_ll_rh_smooth[600:1000])) / 2
    last_10_ll_rh_peaks, _ = mne.preprocessing.peak_finder(last_10_ll_rh_smooth[600:1000], thresh=last_10_ll_rh_thresh)
    last_10_ll_rh_peaks = last_10_ll_rh_peaks + 600

    last_10_ll_lh_thresh = (max(last_10_ll_lh_smooth[600:1000]) - min(last_10_ll_lh_smooth[600:1000])) / 2
    last_10_ll_lh_peaks, _ = mne.preprocessing.peak_finder(last_10_ll_lh_smooth[600:1000], thresh=last_10_ll_lh_thresh)
    last_10_ll_lh_peaks = last_10_ll_lh_peaks + 600

    #Motor Regions
    motor_el_lh_thresh = (max(motor_el_lh_avg_smooth[600:1000]) - min(motor_el_lh_avg_smooth[600:1000])) / 3
    motor_el_lh_peaks, _ = mne.preprocessing.peak_finder(motor_el_lh_avg_smooth[600:1000], thresh=motor_el_lh_thresh)
    motor_el_lh_peaks = motor_el_lh_peaks + 600

    motor_ll_lh_thresh = (max(motor_ll_lh_avg_smooth[600:1000]) - min(motor_ll_lh_avg_smooth[600:1000])) / 3
    motor_ll_lh_peaks, _ = mne.preprocessing.peak_finder(motor_ll_lh_avg_smooth[600:1000], thresh=motor_ll_lh_thresh)
    motor_ll_lh_peaks = motor_ll_lh_peaks + 600

    #Append activations into appropriate lists
    #EL
    el_rh_ad.append(list(el_rh_avg_smooth[el_rh_peaks]))
    el_rh_times_ad.append(list(times[el_rh_peaks]))
    el_lh_ad.append(list(el_lh_avg_smooth[el_lh_peaks]))
    el_lh_times_ad.append(list(times[el_lh_peaks]))

    motor_el_lh_ad.append(list(motor_el_lh_avg_smooth[motor_el_lh_peaks]))
    motor_el_lh_times_ad.append(list(times[motor_el_lh_peaks]))
    #LL
    ll_rh_ad.append(list(ll_rh_avg_smooth[ll_rh_peaks]))
    ll_rh_times_ad.append(list(times[ll_rh_peaks]))
    ll_lh_ad.append(list(ll_lh_avg_smooth[ll_lh_peaks]))
    ll_lh_times_ad.append(list(times[ll_lh_peaks]))

    motor_ll_lh_ad.append(list(motor_ll_lh_avg_smooth[motor_ll_lh_peaks]))
    motor_ll_lh_times_ad.append(list(times[motor_ll_lh_peaks]))
    
    #EL First 10
    el_rh_10_ad.append(list(first_10_el_rh_smooth[first_10_el_rh_peaks]))
    el_rh_10_times_ad.append(list(times[first_10_el_rh_peaks]))
    el_lh_10_ad.append(list(first_10_el_lh_smooth[first_10_el_lh_peaks]))
    el_lh_10_times_ad.append(list(times[first_10_el_lh_peaks]))
    #LL Last 10
    ll_rh_10_ad.append(list(last_10_ll_rh_smooth[last_10_ll_rh_peaks]))
    ll_rh_10_times_ad.append(list(times[last_10_ll_rh_peaks]))
    ll_lh_10_ad.append(list(last_10_ll_lh_smooth[last_10_ll_lh_peaks]))
    ll_lh_10_times_ad.append(list(times[last_10_ll_lh_peaks]))

    # Combined plot for motor and cerebellar time courses
    plt.figure(figsize=(10, 6))

    # Late learning (same color, different line styles)
    plt.plot(times, ll_rh_avg_smooth, label='Cerebellum RH - Late Learning', linestyle='-', color='blue', alpha=0.8)
    plt.plot(times, motor_ll_lh_avg_smooth, label='Motor LH - Late Learning', linestyle='--', color='blue', alpha=0.8)

    # Early learning (same color, different line styles)
    plt.plot(times, el_rh_avg_smooth, label='Cerebellum RH - Early Learning', linestyle='-', color='orange', alpha=0.8)
    plt.plot(times, motor_el_lh_avg_smooth, label='Motor LH - Early Learning', linestyle='--', color='orange', alpha=0.8)

    # Peaks for motor and cerebellar time courses
    plt.scatter(times[motor_ll_lh_peaks], motor_ll_lh_avg_smooth[motor_ll_lh_peaks], color='blue')
    plt.scatter(times[motor_el_lh_peaks], motor_el_lh_avg_smooth[motor_el_lh_peaks], color='orange')

    plt.scatter(times[ll_rh_peaks], ll_rh_avg_smooth[ll_rh_peaks], color='green')
    plt.scatter(times[el_rh_peaks], el_rh_avg_smooth[el_rh_peaks], color='red')

    # General plot settings
    plt.title(f'Adult {s}: Motor and Cerebellar Time Courses')
    plt.xlabel('Time (s)')
    plt.ylabel('Activation (a.u.)')
    plt.legend(loc='best')
    plt.xlim(0, 0.6)  # Adjust time range as needed
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    print('finished adult', {s})
```

```python
i = 0 
for s in child_subs:
    sub = f'vml_meg_{s}'
    subj = f'vml_mri_{s}'
    meg_data = op.join(meg_path, sub, 'vml_meg_%s_tsss_ica_rot_ave.fif' %s)
    meg_data2 = op.join(meg_path, sub, 'vml_meg_%s_tsss_ica_run_ave.fif' %s)

    evoked = mne.read_evokeds(meg_data, proj=True, kind='average', verbose=False) #rot 
    evoked2 = mne.read_evokeds(meg_data2, proj=True, kind='average', verbose=False) #run

    bem_dir = op.join(mri_path, subj, 'bem')
    fname_aseg = op.join(mri_path, subj, 'mri', 'aseg.mgz')
    fname_model = op.join(bem_dir, f"{s}-bem-model.fif")
    fname_bem = op.join(bem_dir, f"{s}-bem-sol.fif")
    fname_trans = op.join(meg_path, sub, f'{sub}-trans.fif')
    fname_fwd = op.join(meg_path, sub, f'{s}-cerebellar-fw-solution-run.fif')
    erm_noise_cov = mne.read_cov(op.join(meg_path, f'vml_meg_{s}', f'{s}_noise-cov.fif'))
    rot_epochs = mne.read_epochs(op.join('VML', 'MEG', sub, f'{sub}_rot_epochs-epo.fif'), verbose=False)
    run_epochs = mne.read_epochs(op.join('VML', 'MEG', sub, f'{sub}_run_epochs-epo.fif'), verbose=False)

    if rot_epochs.baseline == run_epochs.baseline and evoked[0].baseline == evoked2[0].baseline: 
        print (f'Baselines are all the same:{rot_epochs.baseline}')
    else:
        print('Baselines are Different')
        
    el_epos = run_epochs['run2']
    ll_epos = run_epochs['run3']
    fwd = mne.read_forward_solution(fname_fwd, verbose=False)
    src = fwd['src']
    run_filters = mne.beamformer.read_beamformer(op.join(meg_path, sub, f'{s}-run_filters-lcmv.h5'))

    #Early_learning 
    el_stcs = []
    for i in range(len(el_epos)):
        stc = mne.beamformer.apply_lcmv_epochs(el_epos[i], run_filters, return_generator=True, verbose=False)
        el_stcs.append(list(stc))

    #Late Learning
    ll_stcs = []
    for i in range(len(ll_epos)):
        stc1 = mne.beamformer.apply_lcmv_epochs(ll_epos[i], run_filters, return_generator=True, verbose=False)
        ll_stcs.append(list(stc1))

    first_10_el_stcs = el_stcs[:10]
    last_10_ll_stcs = ll_stcs[19:29]

    vol_labels = mne.get_volume_labels_from_src(src, subject=subj, subjects_dir=mri_path)
    surf_labels = mne.read_labels_from_annot(subject=subj, parc='aparc', subjects_dir=mri_path)
    surf_labels = [label for label in surf_labels if label.name == 'postcentral-lh']

    times = el_stcs[0][0].times

    el_rh_tcs = []
    el_lh_tcs = []
    ll_rh_tcs = []
    ll_lh_tcs = []

    motor_el_lh_tcs = []
    motor_ll_lh_tcs = [] 

    for i in range(len(el_stcs)):
        #Extract the timecourses from early/late learning
        el_tcs = mne.extract_label_time_course(el_stcs[i], src=src, labels=vol_labels, mode='max', verbose=False)[:4]

        el_cerebellum_rh = np.mean((el_tcs[0][0], el_tcs[0][1]), axis=0)
        el_cerebellum_lh = np.mean((el_tcs[0][2], el_tcs[0][3]), axis=0)

        el_rh_tcs.append(el_cerebellum_rh)
        el_lh_tcs.append(el_cerebellum_lh)

        motor_el_tc = mne.extract_label_time_course(el_stcs[i], labels=surf_labels, src=src, mode='max')[0]
        motor_el_lh_tcs.append(motor_el_tc)

    for i in range(len(ll_stcs)):

        #Extract the timecourses from early/late learning
        ll_tcs = mne.extract_label_time_course(ll_stcs[i], src=src, labels=vol_labels, mode='max', verbose=False)[:4]

        #Average the Lh/Rh
        ll_cerebellum_rh = np.mean((ll_tcs[0][0], ll_tcs[0][1]), axis=0)
        ll_cerebellum_lh = np.mean((ll_tcs[0][2], ll_tcs[0][3]), axis=0)

        #Keep Track of the timecourses from each epoch 
        ll_rh_tcs.append(ll_cerebellum_rh)
        ll_lh_tcs.append(ll_cerebellum_lh)

        motor_ll_tc = mne.extract_label_time_course(ll_stcs[i], labels=surf_labels, src=src, mode='max')[0]
        motor_ll_lh_tcs.append(motor_ll_tc)

    #RAW MOTOR TCS
    print('EL MOTOR EXTRACTED TCS', 'SHAPE = ', np.shape(motor_el_lh_tcs))
    print('LL MOTOR EXTRACTED TCS', 'SHAPE = ', np.shape(motor_ll_lh_tcs))
    print('CEREBELLUM RH SHAPE', np.shape(el_rh_avg))

    #Shape of data is now (num_epochs x timepoints) (29 x 1701)
    el_rh_avg = np.mean(el_rh_tcs, axis=0)
    el_lh_avg = np.mean(el_lh_tcs, axis=0)
    ll_rh_avg = np.mean(ll_rh_tcs, axis=0)
    ll_lh_avg = np.mean(ll_lh_tcs, axis=0)

    #Motor regions
    motor_el_lh_avg = np.mean(motor_el_lh_tcs, axis=(0,1))
    motor_ll_lh_avg = np.mean(motor_ll_lh_tcs, axis=(0,1))
    motor_el_lh_baseline = np.mean(motor_el_lh_avg[0:200])
    motor_ll_lh_baseline = np.mean(motor_ll_lh_avg[0:200])

    #First 10 el/ last 10 LL
    first_10_el_rh = np.mean(el_rh_tcs[:10], axis=0)
    first_10_el_lh = np.mean(el_lh_tcs[:10], axis=0)
    last_10_ll_rh = np.mean(ll_rh_tcs[19:29], axis=0)
    last_10_ll_lh = np.mean(ll_lh_tcs[19:29], axis=0)

    #Normalize WRT the baseline
    el_rh_baseline = np.mean(el_rh_avg[0:200])
    el_lh_baseline = np.mean(el_lh_avg[0:200])
    ll_rh_baseline = np.mean(ll_rh_avg[0:200])
    ll_lh_baseline = np.mean(ll_lh_avg[0:200])
    b1 = np.mean(first_10_el_rh[0:200])
    b2 = np.mean(first_10_el_lh[0:200])
    b3 = np.mean(last_10_ll_rh[0:200])
    b4 = np.mean(last_10_ll_lh[0:200])

    for i in range(len(el_rh_avg)):
        el_rh_avg[i] = el_rh_avg[i] - el_rh_baseline
        el_lh_avg[i] = el_lh_avg[i] - el_lh_baseline
        ll_rh_avg[i] = ll_rh_avg[i] - ll_rh_baseline
        ll_lh_avg[i] = ll_lh_avg[i] - ll_lh_baseline
        first_10_el_rh[i] = first_10_el_rh[i] - b1
        first_10_el_lh[i] = first_10_el_lh[i] - b2 
        last_10_ll_rh[i] = last_10_ll_rh[i] - b3
        last_10_ll_lh[i] = last_10_ll_lh[i] - b4

    for i in range(len(motor_el_lh_avg)):
        motor_el_lh_avg[i] = motor_el_lh_avg[i] - motor_el_lh_baseline
        motor_ll_lh_avg[i] = motor_ll_lh_avg[i] - motor_ll_lh_baseline

    sigma = 10  # Adjust sigma for more or less smoothing

    #Smooth the TC before finding the peaks: 
    el_rh_avg_smooth = gaussian_filter1d(el_rh_avg, sigma=sigma)
    el_lh_avg_smooth = gaussian_filter1d(el_lh_avg, sigma=sigma)
    ll_rh_avg_smooth = gaussian_filter1d(ll_rh_avg, sigma=sigma)
    ll_lh_avg_smooth = gaussian_filter1d(ll_lh_avg, sigma=sigma)
    first_10_el_rh_smooth = gaussian_filter1d(first_10_el_lh, sigma=sigma)
    first_10_el_lh_smooth = gaussian_filter1d(first_10_el_rh, sigma=sigma)
    last_10_ll_rh_smooth = gaussian_filter1d(last_10_ll_rh, sigma=sigma)
    last_10_ll_lh_smooth = gaussian_filter1d(last_10_ll_lh, sigma=sigma)

    motor_el_lh_avg_smooth = gaussian_filter1d(motor_el_lh_avg, sigma=sigma)
    motor_ll_lh_avg_smooth = gaussian_filter1d(motor_ll_lh_avg, sigma=sigma)

    # Compute dynamic thresholds and find peaks
    el_rh_thresh = (max(el_rh_avg_smooth[600:1000]) - min(el_rh_avg_smooth[600:1000])) / 3
    el_rh_peaks, _ = mne.preprocessing.peak_finder(el_rh_avg_smooth[600:1000], thresh=el_rh_thresh)
    el_rh_peaks = el_rh_peaks + 600

    el_lh_thresh = (max(el_lh_avg_smooth[600:1000]) - min(el_lh_avg_smooth[600:1000])) / 3
    el_lh_peaks, _ = mne.preprocessing.peak_finder(el_lh_avg_smooth[600:1000], thresh=el_lh_thresh)
    el_lh_peaks = el_lh_peaks + 600

    ll_rh_thresh = (max(ll_rh_avg_smooth[600:1000]) - min(ll_rh_avg_smooth[600:1000])) / 3
    ll_rh_peaks, _ = mne.preprocessing.peak_finder(ll_rh_avg_smooth[600:1000], thresh=ll_rh_thresh)
    ll_rh_peaks = ll_rh_peaks + 600

    ll_lh_thresh = (max(ll_lh_avg_smooth[600:1000]) - min(ll_lh_avg_smooth[600:1000])) / 3
    ll_lh_peaks, _ = mne.preprocessing.peak_finder(ll_lh_avg_smooth[600:1000], thresh=ll_lh_thresh)
    ll_lh_peaks = ll_lh_peaks + 600

    first_10_el_rh_thresh = (max(first_10_el_rh_smooth[600:1000]) - min(first_10_el_rh_smooth[600:1000])) / 3
    first_10_el_rh_peaks, _ = mne.preprocessing.peak_finder(first_10_el_rh_smooth[600:1000], thresh=first_10_el_rh_thresh)
    first_10_el_rh_peaks = first_10_el_rh_peaks + 600

    first_10_el_lh_thresh = (max(first_10_el_lh_smooth[600:1000]) - min(first_10_el_lh_smooth[600:1000])) / 3
    first_10_el_lh_peaks, _ = mne.preprocessing.peak_finder(first_10_el_lh_smooth[600:1000], thresh=first_10_el_lh_thresh)
    first_10_el_lh_peaks = first_10_el_lh_peaks + 600

    last_10_ll_rh_thresh = (max(last_10_ll_rh_smooth[600:1000]) - min(last_10_ll_rh_smooth[600:1000])) / 3
    last_10_ll_rh_peaks, _ = mne.preprocessing.peak_finder(last_10_ll_rh_smooth[600:1000], thresh=last_10_ll_rh_thresh)
    last_10_ll_rh_peaks = last_10_ll_rh_peaks + 600

    last_10_ll_lh_thresh = (max(last_10_ll_lh_smooth[600:1000]) - min(last_10_ll_lh_smooth[600:1000])) / 3
    last_10_ll_lh_peaks, _ = mne.preprocessing.peak_finder(last_10_ll_lh_smooth[600:1000], thresh=last_10_ll_lh_thresh)
    last_10_ll_lh_peaks = last_10_ll_lh_peaks + 600

    #Motor Regions
    motor_el_lh_thresh = (max(motor_el_lh_avg_smooth[600:1000]) - min(motor_el_lh_avg_smooth[600:1000])) / 3
    motor_el_lh_peaks, _ = mne.preprocessing.peak_finder(motor_el_lh_avg_smooth[600:1000], thresh=motor_el_lh_thresh)
    motor_el_lh_peaks = motor_el_lh_peaks + 600

    motor_ll_lh_thresh = (max(motor_ll_lh_avg_smooth[600:1000]) - min(motor_ll_lh_avg_smooth[600:1000])) / 3
    motor_ll_lh_peaks, _ = mne.preprocessing.peak_finder(motor_ll_lh_avg_smooth[600:1000], thresh=motor_ll_lh_thresh)
    motor_ll_lh_peaks = motor_ll_lh_peaks + 600

    #Append activations into appropriate lists
    #EL
    el_rh.append(list(el_rh_avg_smooth[el_rh_peaks]))
    el_rh_times.append(list(times[el_rh_peaks]))
    el_lh.append(list(el_lh_avg_smooth[el_lh_peaks]))
    el_lh_times.append(list(times[el_lh_peaks]))

    motor_el_lh.append(list(motor_el_lh_avg_smooth[motor_el_lh_peaks]))
    motor_el_lh_times.append(list(times[motor_el_lh_peaks]))
    #LL
    ll_rh.append(list(ll_rh_avg_smooth[ll_rh_peaks]))
    ll_rh_times.append(list(times[ll_rh_peaks]))
    ll_lh.append(list(ll_lh_avg_smooth[ll_lh_peaks]))
    ll_lh_times.append(list(times[ll_lh_peaks]))

    motor_ll_lh.append(list(motor_ll_lh_avg_smooth[motor_ll_lh_peaks]))
    motor_ll_lh_times.append(list(times[motor_ll_lh_peaks]))

    #EL First 10
    el_rh_10.append(list(first_10_el_rh_smooth[first_10_el_rh_peaks]))
    el_rh_10_times.append(list(times[first_10_el_rh_peaks]))
    el_lh_10.append(list(first_10_el_lh_smooth[first_10_el_lh_peaks]))
    el_lh_10_times.append(list(times[first_10_el_lh_peaks]))
    #LL Last 10
    ll_rh_10.append(list(last_10_ll_rh_smooth[last_10_ll_rh_peaks]))
    ll_rh_10_times.append(list(times[last_10_ll_rh_peaks]))
    ll_lh_10.append(list(last_10_ll_lh_smooth[last_10_ll_lh_peaks]))
    ll_lh_10_times.append(list(times[last_10_ll_lh_peaks]))

    # Combined plot for motor and cerebellar time courses
    plt.figure(figsize=(10, 6))

    # Late learning (same color, different line styles)
    plt.plot(times, ll_rh_avg_smooth, label='Cerebellum RH - Late Learning', linestyle='-', color='blue', alpha=0.8)
    plt.plot(times, motor_ll_lh_avg_smooth, label='Motor LH - Late Learning', linestyle='--', color='blue', alpha=0.8)

    # Early learning (same color, different line styles)
    plt.plot(times, el_rh_avg_smooth, label='Cerebellum RH - Early Learning', linestyle='-', color='orange', alpha=0.8)
    plt.plot(times, motor_el_lh_avg_smooth, label='Motor LH - Early Learning', linestyle='--', color='orange', alpha=0.8)

    # Peaks for motor and cerebellar time courses
    plt.scatter(times[motor_ll_lh_peaks], motor_ll_lh_avg_smooth[motor_ll_lh_peaks], color='blue')
    plt.scatter(times[motor_el_lh_peaks], motor_el_lh_avg_smooth[motor_el_lh_peaks], color='orange')

    plt.scatter(times[ll_rh_peaks], ll_rh_avg_smooth[ll_rh_peaks], color='green')
    plt.scatter(times[el_rh_peaks], el_rh_avg_smooth[el_rh_peaks], color='red')

    # General plot settings
    plt.title(f'Child {s}: Motor and Cerebellar Time Courses')
    plt.xlabel('Time (s)')
    plt.ylabel('Activation (a.u.)')
    plt.legend(loc='best')
    plt.xlim(0, 0.6)  # Adjust time range as needed
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    print('finished child', {s})
```

```python
import numpy as np
import mne
from scipy.signal import welch
import os.path as op

stmp = '030'

# Define parameters
stc = mne.read_source_estimate(op.join(meg_path, f'vml_meg_{stmp}', f'morphed_DS_{stmp}_cond-run 2 -stc.h5'))
labels_motor = ["precentral-lh", "precentral-rh", "postcentral-lh", "postcentral-rh"]
labels_cerebellum = ["Cerebellum-lh", "Cerebellum-rh"] 
fmin, fmax = 1, 40  # Frequency range for PSD (adjust as needed)
sfreq = 1000  # Adjust to your sampling frequency

# Load the labels
labels = mne.read_labels_from_annot(f'vml_mri_{stmp}', parc='aparc', subjects_dir=mri_path)

# Function to extract PSD
def compute_psd(data, sfreq, fmin, fmax):
    freqs, psd = welch(data, sfreq, nperseg=sfreq//2, axis=1)  # Welch's method
    freq_mask = (freqs >= fmin) & (freqs <= fmax)
    return np.mean(psd[:, freq_mask], axis=1)  # Average PSD in the range

# Extract motor and cerebellar region indices
motor_indices = np.concatenate([stc.in_label(label).vertices for label in labels if label.name in labels_motor])
cerebellar_indices = np.concatenate([stc.in_label(label).vertices for label in labels if label.name in labels_cerebellum])

# Extract time series data for these regions
motor_data = stc.data[motor_indices]
cerebellar_data = stc.data[cerebellar_indices]

# Compute PSD
motor_psd = compute_psd(motor_data, sfreq, fmin, fmax)
cerebellar_psd = compute_psd(cerebellar_data, sfreq, fmin, fmax)

# Create a new STC file with PSD values
psd_data = np.zeros_like(stc.data)
psd_data[motor_indices] = motor_psd[:, None]  # Broadcast to match time shape
psd_data[cerebellar_indices] = cerebellar_psd[:, None]

# Save the new STC file
stc_psd = mne.SourceEstimate(psd_data, vertices=stc.vertices, tmin=stc.tmin, tstep=stc.tstep, subject=stc.subject)
stc_psd.save("TESTEST")  # Replace with your output filename
```
