```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
file_path = "VML_FINAL_DATA2.csv"
df = pd.read_csv(file_path, sep=",")
print(df.head())
```

```python
import matplotlib.pyplot as plt

# Ensure df is loaded from the previous cell (cell 2)
# Filter for subject 7
df_subject7 = df[df['sub'] == 7]

# Group by run for subject 7
grouped_subject7 = df_subject7.groupby(['run'])

# Create a single figure for all plots for subject 7
plt.figure(figsize=(12, 8))

if grouped_subject7.ngroups == 0:
    print("No data found for subject 7.")
else:
    # Plot absolute ErrorAngle against trial for each run of subject 7 on the same figure
    for run, group_df in grouped_subject7:
        plt.plot(group_df['trial'], group_df['ErrorAngle'].abs(), marker='o', linestyle='-', label=f'Subject 7, Run {run}')

    # Setup plot details once after all lines are plotted
    plt.title('Absolute Error Angle vs. Trial for Subject 7 (All Runs)')
    plt.xlabel('Trial')
    plt.ylabel('Absolute Error Angle')
    plt.grid(True)
    plt.legend()
    plt.show() # Display the plot once with all lines
```
